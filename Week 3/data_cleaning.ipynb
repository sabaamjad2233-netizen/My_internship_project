```python
df = pd.read_csv('../forum_posts.csv')
df.head()
```

```python
def clean_my_text(text):
    if isinstance(text, str):
        # 1. HTML Tags hatao (jaise <p>, <div>)
        text = re.sub(r'<[^>]+>', '', text)
        
        # 2. Internet ke links hatao (jaise http, www)
        text = re.sub(r'https?://\S+|www\.\S+', '', text)
        
        # 3. Emojis hatao
        text = text.encode('ascii', 'ignore').decode('ascii')
        
        # 4. Faltu khali jagah saaf karo
        text = " ".join(text.split())
        
        return text
    return text
```

```python
# Hum computer ko bol rahe hain ke 'Text' column par safai chalao aur nayi column 'clean_text' banao
df['clean_text'] = df['Text'].apply(clean_my_text)

# Yeh dekhne ke liye ke safai hui ya nahi
df[['Text', 'clean_text']].head()
```

```python
# 1. Pehle agar koi row bilkul khali hai to usko hatao
df = df[df['clean_text'].notna()]

# 2. Sirf wo rows rakho jahan words ki tadad 5 ya us se zyada ho
df = df[df['clean_text'].apply(lambda x: len(str(x).split()) >= 5)]

# 3. Check karne ke liye ke ab kitna data bacha hai
print("Baqi bacha hua data:", df.shape)
```

```python
# Hum computer ko bol rahe hain ke is saaf data ko 'clean_dataset.csv' ke naam se save kar do
df.to_csv('clean_dataset.csv', index=False)

print("Mubarak ho Saba! Saaf data save ho gaya hai.")
```
