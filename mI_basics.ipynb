```python

df = pd.read_csv('../Week 3/clean_dataset.csv')


df.head()
```

```python
import pandas as pd
import numpy as np
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import make_pipeline

training_data = {
    'text': [
        "I love this, it is absolutely amazing and wonderful!",
        "This is the worst experience ever, very bad and poor.",
        "Great job, highly recommended and extremely happy.",
        "I hate this, so disappointing and useless.",
        "Beautiful day, full of joy and positive vibes.",
        "Terrible service, completely waste of time and horrible."
    ],
    'sentiment': ['Positive', 'Negative', 'Positive', 'Negative', 'Positive', 'Negative']
}
train_df = pd.DataFrame(training_data)

model_pipeline = make_pipeline(
    TfidfVectorizer(),
    LogisticRegression()
)

model_pipeline.fit(train_df['text'], train_df['sentiment'])

posts = df['clean_text'].fillna("").tolist()

predictions = model_pipeline.predict(posts)
probabilities = model_pipeline.predict_proba(posts)


confidence_scores = [f"{max(prob) * 100:.2f}%" for prob in probabilities]

output_df = pd.DataFrame({
    'post': posts,
    'sentiment': predictions,
    'confidence': confidence_scores
})

output_df.to_csv('sentiment_predictions.csv', index=False)

pd.set_option('display.max_rows', None)
output_df
```
